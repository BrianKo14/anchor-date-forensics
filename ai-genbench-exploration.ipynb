## Fake image dataset

```python
import datasets, itertools
sys.path.insert(0, "dataset_creation")
from custom_features.large_image import LargeImage
datasets.features.features.register_feature(LargeImage, "LargeImage")

from datasets import load_dataset

ds = load_dataset("lrzpellegrini/AI-GenBench-fake_part", streaming=True)
print(ds)  # just prints split names, no download triggered

# The dataset is ~180K images, so for download efficiency:
# we shuffle over a rolling buffer so the peek isn't all one generator,
# then we take a small number of examples
peek = ds["train"].shuffle(seed=42, buffer_size=500)
sample = list(itertools.islice(peek, 6))
for ex in sample:
    print(ex["generator"], ex["image"].size)
```

```python
fig, axes = plt.subplots(2, 3, figsize=(12, 8))
for ax, ex in zip(axes.flat, sample):   # 'sample' comes from Cell 4
    ax.imshow(ex["image"])
    ax.set_title(f'{ex["generator"]}\n{ex["width"]}×{ex["height"]}', fontsize=9)
    ax.axis("off")
plt.tight_layout()
```
